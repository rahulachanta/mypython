Import TensorFlow and Keras utilities into the notebook


```python
import tensorflow as tf
import tensorflow.keras
import numpy as np
```

```python
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from tensorflow.keras.preprocessing import image
from tensorflow.keras.applications.vgg16 import VGG16, preprocess_input, decode_predictions
from tensorflow.keras.applications.resnet50 import ResNet50, preprocess_input

```

```python
#load the model into the notebook
#model = VGG16()
model = ResNet50()
```

```python
#load an image into the notebook
#Step 1: Load image (224 × 224 × 3)
#img = load_img('Kitten1.jpg', target_size=(224,224), color_mode="rgb")
img = load_img('Kitten2.png', target_size=(224,224), color_mode="rgb")
img
```

```python
#prepare the image for the model
#Step 2: Convert to NumPy array
img_array = image.img_to_array(img)
#print(img_array)
print(img_array.shape)
```

```python
#make the prediction
##result = model.predict(image)
#label = decode_predictions(result)
#label = label[0][0]
```

```python
#Step 3: Add batch dimension (MOST IMPORTANT)
img_array = np.expand_dims(img_array, axis=0)
print(img_array.shape)
```

```python
#Step 4: Preprocess for VGG16

from tensorflow.keras.applications.vgg16 import preprocess_input

img_array = preprocess_input(img_array)
```

```python
#Step 5: Predict
result = model.predict(img_array)
label = decode_predictions(result)
label = label[0][0]
```

```python
#Display the classification 
print('%s (%.2f%%)' % (label[1], label[2] * 100))
```
