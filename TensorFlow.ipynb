```python
import tensorflow as tf
```

```python
h = print("Hello World")
```

```python
h = tf.constant("Hello World")
tf.print(h)
```

```python
a = tf.constant("Hello There")
b = 9+70
c = tf.Variable(300)
tf.print(a)
tf.print(b)
tf.print(c)
```

```python
a = tf.constant(6)
b = tf.Variable(6)
b = tf.Variable(2)
c = a*b
tf.print(c)
```

```python
x = 7
y = 8
z = 9

@tf.function
def result(x,y,z):
    a = x + y -z
    return a 

b = result(x,y,z)
tf.print(b)

b = result(1,7,3)
tf.print(b)
```

```python
#ragged tensors
animals = tf.ragged.constant([['c','a','t'], 
                              ['h', 'i','p','p','o','t','a','m','u','s'],
                              ['b','u','f','f','a','l','o']])
print(animals[2])
```
